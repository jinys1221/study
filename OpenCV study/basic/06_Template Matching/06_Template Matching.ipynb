```python
# 라이브러리
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## 1. 이미지와 템플릿 이미지 불러오기

```python
img = cv2.imread('D:/study/OpenCV study/06_Template Matching/Original.png')  # 원본 이미지

# 템플릿으로 사용할 부분 자르기 (x, y 좌표 and w, h 크기)
x, y, w, h = 320, 120, 320, 100  # 자르고자 하는 영역의 좌표와 크기
template = img[y:y+h, x:x+w]

# 템플릿 이미지 저장
cv2.imwrite('D:/study/OpenCV study/06_Template Matching/template_image.png', template)
```

```python
# 템플릿 크기와 원본 이미지 크기 확인
w, h = template.shape[1], template.shape[0]

# 템플릿 매칭 수행 (일치 정도를 점수로 계산)
result = cv2.matchTemplate(img, template, cv2.TM_CCOEFF_NORMED)

# 가장 일치하는 위치 찾기 (최대 값 위치)
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

# 최적의 매칭 위치 구하기 (최대 값 위치)
top_left = max_loc
bottom_right = (top_left[0] + w, top_left[1] + h)

# 매칭된 영역에 사각형 그리기
cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 2)

# 결과 이미지 출력
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show() 
```
