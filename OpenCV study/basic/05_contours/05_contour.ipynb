```python
# 라이브러리
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## 1. 이미지 불러오기

```python
img = cv2.imread("D:/study/OpenCV study/image/wall.jpg", cv2.IMREAD_GRAYSCALE)
```

## 2. 이진화 및 엣지 검출

```python
# 이진화
_, threshold_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# Canny 검출하기 전 블러 처리
Blur = cv2.GaussianBlur(img, (7, 7), 0)
# Canny 엣지 검출
Edge = cv2.Canny(Blur, 40, 100)
```

## 3. 윤곽선 검출
- `cv2.findContours()`: 이진화된 이미지에서 윤곽선을 검출하는 함수
- `cv2.drawContours()`: 검출된 윤곽선을 이미지에 시각화하는 함수

```python
# 윤곽선 찾기
contours, hierarchy = cv2.findContours(threshold_img, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# 윤곽선 원본 이미지에 그리기
img_contours = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
Draw = cv2.drawContours(img_contours, contours, -1, (0, 0, 255), 4)
```
