```python
# 라이브러리
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## 1. 이미지 불러오기

```python
img = cv2.imread('D:/study/OpenCV study/image/opencv-logo-white.png')  # 원본 이미지
```

## 2. 윤곽선 검출

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
```

## 3. 각 윤곽선마다 도형 그리기

```python
for cnt in contours:
    cv2.drawContours(img, [cnt], 0, (0, 255, 0), 2)
    x, y, w, h = cv2.boundingRect(cnt)   # x, y 왼쪽 위 꼭짓점 좌표, w, h : 너비, 높이
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 1)
```

## 4. 시각화

```python
# BGR → RGB 변환
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 시각화
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title('Contour Result')
plt.axis('off')
plt.show()
```

- `cv2.drawContours`: img내의 윤곽선따라 초록선을 그려주고 있다.
- `cv2.boundingRect`: 주어진 점을 감싸는 최소 크기 사각형을 반환
