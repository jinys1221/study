```python
import cv2
import matplotlib.pyplot as plt
```

## 1. 이미지 불러오기

```python
img_color = cv2.imread('D:\study\OpenCV study\image\hand.jpg')
img_gray = cv2.imread('D:\study\OpenCV study\image\hand.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
rgb = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)  

# 시각화
plt.imshow(rgb, cmap="gray")
plt.title("hand")
plt.axis("off")
plt.show()
```

## 2. 히스토그램 계산

```python
# 그레이스케일
hist_gray = cv2.calcHist([img_gray], [0], None, [256], [0, 256])
```

- `[img_gray]`: 입력 이미지 리스트 형태
- `[0]`: 채널 번호
- `None`: 마스크 사용 안 함
- `[256]`: 빈 개수 (0~255)
- `[0, 256]`: 픽셀 값 범위

```python
# 컬러 채널별
colors = ('b', 'g', 'r')
hist_colors = []
for i, col in enumerate(colors):
    hist = cv2.calcHist([img_color], [i], None, [256], [0, 256])
    hist_colors. append((hist, col))
```

## 3. 히스토그램 시각화

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Grayscale Histogram')
plt.plot(hist_gray, color='black')

plt.subplot(1, 2, 2)
plt.title('Color Histogram')
for hist, col in hist_colors:
    plt.plot(hist, color=col)
plt.xlim([0, 256])

plt.show()
```

원본 사진이 밝은 부분보다 어두운 부분이 훨씬 많아서 0 부근으로 크게 치우친 것을 확인할 수 있다.
